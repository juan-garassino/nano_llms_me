## Setup
First, we ensure we are in the project root and import the necessary modules.

```python
import os
import sys
import torch

# Ensure project root is in path
project_root = os.path.abspath(os.path.join(os.getcwd(), ".."))
if project_root not in sys.path:
    sys.path.append(project_root)

from nano_moe.main import main
from nano_moe.config import TrainingConfig

print(f"Project Root: {project_root}")
print(f"Device: {'cuda' if torch.cuda.is_available() else 'cpu'}")
```

## 1. Universal SFPT (Text Generation)
The Universal SFPT uses Sparse Fourier layers to process sequences. We'll train it on Shakespeare.

```python
cfg_universal = TrainingConfig()
cfg_universal.model_type = "universal_phase"
cfg_universal.dataset_type = "text"
cfg_universal.dataset_names = ["tinyshakespeare"]
cfg_universal.batch_size = 4  # Adjust based on VRAM
cfg_universal.seq_len = 64
cfg_universal.epochs = 1
cfg_universal.save_dir = "training_outputs/universal_sfpt"

# Run Training
main(cfg_universal)
```

## 2. True Holographic SFPT (Complex-Valued)
This model uses explicit complex numbers and phase rotations. It simulates optical interference.

```python
cfg_holographic = TrainingConfig()
cfg_holographic.model_type = "true_phase"
cfg_holographic.dataset_type = "text"
cfg_holographic.dataset_names = ["tinyshakespeare"]
cfg_holographic.batch_size = 4
cfg_holographic.seq_len = 64
cfg_holographic.epochs = 1
cfg_holographic.save_dir = "training_outputs/true_holographic"

# Run Training
main(cfg_holographic)
```
